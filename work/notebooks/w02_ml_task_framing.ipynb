## 1. My lane as an ML task (type)

**Task Type**: Ranking / scoring

**Why**: The goal is to improve the decision of "Which decaying/underperforming page should an editor update today to maximize recovered traffic?". The output supports editors by providing a prioritized list of pages to act upon. A wrong call costs wasted editor hours and missed opportunities to recover traffic on pages that actually needed a refresh.


```python
print("Task Type: Ranking / Scoring")
```

## 2. Target or proxy

**Target**: We will predict the expected `ctr` (Click-Through Rate) based on content features (like `avg_position`, `word_count`, `content_type`). We are specifically avoiding the `is_declining_label` provided in the raw data because it is defined by a rule (`trend_pct < -0.15`), and our model would just learn the rule rather than the world. By predicting an expected, observed continuous metric like `ctr`, we can compare it to the actual `ctr` to identify underperforming pages (opportunities for a refresh).

```python
print("Target: Expected CTR (observed metric)")
```

## 3. Success metric

**Success Metric**: Precision@K (e.g., Precision@50).
Since the editors can only update a limited number of pages, we care about the quality of the top recommendations. If we recommend the top 50 pages to fix, how many of them were true underperformers that successfully recovered traffic after the intervention?

```python
print("Metric: Precision@K")
```

## 5. Why ML beats a fixed rule here

A plain rule (e.g., `ctr < 2% and impressions > 1000`) is too rigid. `ctr` naturally varies heavily by `avg_position` (position 1 vs position 10), `content_type`, and seasonal trends. The pattern is too messy to write by hand because we need to model the non-linear interactions between position, word count, type, and historical performance to find the true anomalies, rather than just applying a flat threshold.

```python
print("ML handles non-linear interactions across many features.")
```
